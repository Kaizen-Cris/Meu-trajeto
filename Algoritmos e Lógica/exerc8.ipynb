```python
## Construa um programa que calcule:
#  Tenho uma lista de preços de produtos, se fizermos um reajuste de 5% a mais em todos os produtos, como ficam os preços dos produtos?


# Dados da minha Lista
lista_precos = [50, 59.99, 100, 110, 130, 140]

# Reajuste de valores
reajuste = 0.05

# Lista reajustada
lista_atualizada = []

## Formula para pegar cada item da lista individualmente e aplicar o reajuste
#  O preco do for representa cada item da lista_precos e o append serve para adicionar o novo valor reajustado
for preco in lista_precos:
    novo_preco = preco * (1 + reajuste)
    lista_atualizada.append(round(novo_preco,2))

print(lista_atualizada)    
```
