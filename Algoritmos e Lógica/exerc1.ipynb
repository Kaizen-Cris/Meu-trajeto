### Construa um programa que calcule: Quanto devemos cobrar em um projeto de programacao se trabalhamos 8h por dia, demoramos 15 dias para fazer o projeto e cobramos R$100,00/h?



```python
# Descobrindo a quantidade de horas trabalhadas totais
horas_trabalhadas = 8
dias_no_projeto = 15
horas_dia_trabalhado = horas_trabalhadas * dias_no_projeto

# Achando o valor pelo total de horas trabalhadas
preço_hora = 100
valor_final = horas_dia_trabalhado * preço_hora

# Valor cobrado no projeto
print(valor_final)
```
