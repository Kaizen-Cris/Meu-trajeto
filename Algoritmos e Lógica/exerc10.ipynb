```python
## Construa um programa para:
# -Calcular o salário dos funcionários com reajuste de 10% (a partir de uma lista de salário)
# -Calcular o custo total de salários que o RH vai ter que pagar
# -Calcular a diferença em R$ entre o cenário anterior e o cenário atual

# Lista de salário dos funcionarios
lista_salarios = [1800, 2000, 1500, 3000, 4500]
custo_total = sum(lista_salarios)

# Reajuste
reajuste = 0.1

# Funcao calculo para definir o salário já com o reajuste
def calcular_lista_salarios_reajustados(lista_salarios, reajuste):
    lista_salario_reajustada = []
    for salario in lista_salarios:
        novo_salario = salario * (1 + reajuste)
        lista_salario_reajustada.append(round(novo_salario,2))
    return lista_salario_reajustada


# Funcao custo de salário total
def calcular_custo_total(lista_salarios):
    total = sum(lista_salarios)
    return total

# Funcao para descobrir a diferença antes e depois do reajuste
def calcular_diferenca(novo_custo_total, custo_total):
    diferenca = novo_custo_total - custo_total
    print(f"O custo adicional de salários por conta do reajuste salarial é de R$ {diferenca:.2f}")

# Calcula a lista reajustada e obtém o total reajustado
lista_salario_reajustada = calcular_lista_salarios_reajustados(lista_salarios, reajuste)
novo_custo_total = calcular_custo_total(lista_salario_reajustada)

# Exibe a lista reajustada
print(f"Lista de salários reajustada: {lista_salario_reajustada}")

# Exibe o custo total atual
print(f"O custo que o RH tem com salários atualmente é de R$ {custo_total:.2f}")

# Calcula e exibe a diferença nos custos
calcular_diferenca(novo_custo_total, custo_total)
```
