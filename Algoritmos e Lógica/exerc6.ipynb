```python
## Construa um programa que calcule: Qual seria o salario de um funcionário após 10 anos
## se todo ano ele ganhasse 10% de aumento, salario inicial de R$ 2000.00

# Salario inicial
salario = 2000.00

# Anos trabalhado
tempo_trabalho = 10

# Aumento anual
aumento = 0.1

# Formula para calcular os 10% de aumento anual, foi usado o range para definir quantas vezes, e o for como loop para fazer o calculo
# fazendo a variavel salario ser atualizada a cada volta, e formatando o valor final com "".2f" para considerar somente 2 casas decimais

for n in range(tempo_trabalho):
    salario = salario * (1 + aumento)
    
    print(f"Ano {n+1}: R$ {salario:.2f}")

# Salario atualizado após 10 anos de reajustes anuais
print(f"Após 10 anos de trabalho, meu salário recebido seria de R$ {salario:.2f}")
```
