## Construa um programa que calcule: Quanto custa encher o tanque de um carro que tem 50 litros de capacidade,
## está com 20 litros de combustivel atualmente e o custo do combustivel é de R$ 5,80




```python
# Descobri quantos litros preciso para completar o tanque
capacidade_tanque = 50
combustivel_restante = 20
necessidade_de_combustivel = capacidade_tanque - combustivel_restante

# Descobri o custo para completar o tanque, fazendo a necessidade_de_combustivel * o valor do combustivel
preco_combustivel = 5.80
custo_para_encher_tanque = necessidade_de_combustivel * preco_combustivel

# Custo para encher o tanque
print(custo_para_encher_tanque)
```
