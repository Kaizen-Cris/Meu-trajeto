```python
## Construa uma programa que calcule: Qual dever ser o bônus de um funcionário?
## Se a empresa bateu a meta de 10000 vendas e se ele vendeu mais de 1000 unidades, o bonus tem que ser de R$ 250,00, caso contrário, o bônus tem que ser R$ 50,00

# Input para perguntar as unidades vendidas pelo funcionário
qtd_vendida_func = int(input("Quantas unidades o funcionário vendeu?:"))
qtd_vendida_empresa = int(input("Quantas vendas a empresa fez?:"))

# Meta de vendas para receber o bônus
meta_func = 1000
meta_empresa = 10000

# Bônus por meta batida
bonus = 250.00

# formula para identificar quanto de bonus ele deve ganhar e mostrar o resultado

if qtd_vendida_func > meta_func and qtd_vendida_empresa > meta_empresa:
    print(f"O funcionário vendeu {qtd_vendida_func} unidades e a empresa bateu a meta de {meta_empresa} vendas, ele receberá o bônus de R$ {bonus}")
else:
    bonus = 50.00
    print(f"O funcionário ou a empresa nao atingiram a meta, ele receberá o bônus de R$ {bonus}")
```
