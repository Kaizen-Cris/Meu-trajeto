```python
## Construa um programa que calcule: Calcule o contrário agora, 
#  quanto tempo demora para esse funcionario chegar em um salário de 10000 reais

# Salario inicial
salario = 2000.00

# Meta de salário
meta_salario = 10000.00

# Aumento anual
aumento = 0.1

# Contador de anos
anos = 0

# Formula para descobrir quantos anos demoraria para o meu salario chegar em 10000 enquanto ele faz a conta
while salario < meta_salario:
    salario = (salario * aumento) + salario
    anos = anos + 1

# Resposta de quantos anos demoraria para o salario mensal ficar acima de R$ 10000.00
print(f"Após {anos} anos de trabalho, meu salário mensal seria R$ {salario:.2f}")
```
