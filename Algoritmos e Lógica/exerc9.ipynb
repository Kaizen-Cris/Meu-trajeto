```python
## Construa um programa que calcule:
#  Voce tem uma lista de preços de produtos, todos os produtos acima de R$ 5000,00 vao ser reajustados em 5% e todos abaixo de R$ 5000,00 vao ser reajustados em 10%
#  como ficam os preços dos produtos


# Dados da minha Lista
lista_precos = [500, 590.99, 1000, 1100, 13000, 14000]

# Reajuste de valores
reajuste1 = 0.05
reajuste2 = 0.1

# Lista reajustada
lista_atualizada = []

## Formula para pegar cada item da lista individualmente e aplicar o reajuste
#  O preco do for representa cada item da lista_precos e o append serve para adicionar o novo valor reajustado
for preco in lista_precos:
    if preco > 5000:
        novo_preco = preco * (1 + reajuste1)
    else:
        novo_preco = preco * (1 + reajuste2)
    lista_atualizada.append(round(novo_preco,2))

print(lista_atualizada)    
```
