#### 8. Vamos criar um conversor de temperatura. Faça um Programa que peça a temperatura em graus Fahrenheit, transforme e mostre a temperatura em graus Celsius.
$C = \frac{5}{9}(F-32)$

```python
fahrenheit = float(input("informe a temperatura em Graus Fahrenheit: "))

c = 5/9 * (fahrenheit - 32)

print(f"A temperatura informada é igual á {c:.2f} graus Celsius")
```

#### 9. Faça um Programa que peça a temperatura em graus Celsius, transforme e mostre em graus Fahrenheit.
$F = \frac{9}{5}C + 32$

```python
celsius = float(input("Informe a temperatura em Graus Celsius"))

f = (celsius * 9/5) + 32

print(f"A temperatura informada é igual á {f} graus Fahrenheit")
```

#### 10. Tendo como dados de entrada a altura (h) de uma pessoa, construa um algoritmo que calcule seu peso ideal, usando a seguinte fórmula:
$P = 72,7h - 58$

#### 11. Tendo como dado de entrada a altura (h) de uma pessoa, construa um algoritmo que calcule seu peso ideal, utilizando as seguintes fórmulas:
##### a. Para homens: $P = 72,7h - 58$
##### b. Para mulheres: $P = 62,1h - 44,7$

```python
altura = float(input("Informe a sua altura: "))
sexo = int(input("Informe seu sexo considerando 1 para Masculino e 2 para Feminino: "))

peso_ideal = ((72.7 * altura) - 58) if sexo == 1 else ((62.1 * altura) - 44.7)

print(f"O seu peso ideal de acordo com o seu sexo é de {peso_ideal}Kg")
```

#### 12. Faça um Programa que pergunte quanto você ganha por hora e o número de horas trabalhadas no mês.

```python
vh = float(input("Quanto voce ganha por hora?"))
ht = float(input("Quantas horas voce trabalhou"))

print(vh)
print(ht)
```

#####  Calcule o salário bruto (horas * salario por hora)

```python
salario_bruto = vh * ht
print(f"salario bruto: R$ {salario_bruto: ,.2f}")
```

##### Calcule o desconto do IR (11% do salário bruto)

```python
ir = salario_bruto * 0.11

print(f"O IR é R$ {ir: ,.2f}")
```

##### Calcule o desconto do INSS (8% do salário bruto)

```python
inss = salario_bruto * 0.08

print(f"O INSS é R$ {inss: ,.2f}")
```

##### Calcule o desconto do sindicato (5% do salário bruto)

```python
sindicato = salario_bruto * 0.05

print(f"O desconto do Sindicato é R$ {sindicato: ,.2f}")
```

##### Calcule o salário líquido (salário bruto - descontos)

```python
descontos = ir + inss + sindicato

print(f"O salário liquido é R$ {salario_bruto - descontos: ,.2f}")
```

#### 14. Faça um programa que peça o tamanho de um arquivo para download (em MB) e a velocidade de um link de Internet (em Mbps), calcule e informe o tempo aproximado de download do arquivo usando este link (em minutos).

Detalhe: MB significa megabyte, Mb (com b minúsculo) significa megabit. Um megabit é 1/8 de um megabyte. 

```python
arquivo = float(input("Qual o tamanho do arquivo?: "))
vel_net = float(input("Qual a velocidade da sua internet?: "))

tamanho = arquivo * 8

tempo = tamanho / vel_net

tempo_minutos = tempo / 60

print(f"Levará {tempo_minutos} minutos")
```
