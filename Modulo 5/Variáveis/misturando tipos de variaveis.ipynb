#### Método Format

```python
faturamento = 1000
custo = 500
lucro = faturamento - custo

print("O faturamento da loja foi {}. O custo da loja foi {}. Assim, o lucro da loja foi de {} reais.".format(faturamento, custo, lucro))
```

#### F-string

```python
faturamento = 1000
custo = 400
lucro = faturamento - custo

# com format
print("O faturamento da loja foi {}. O custo da loja foi {}. Assim, o lucro da loja foi de {} reais.".format(faturamento, custo, lucro))

# com f-string
print(f"O faturamento da loja foi {faturamento}. O custo da loja foi {custo}. Assim, o lucro da loja foi de {lucro} reais.")
```
